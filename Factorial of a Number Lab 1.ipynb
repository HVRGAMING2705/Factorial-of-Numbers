```python
import numpy as np

def get_factorial():
    print("\n--- NumPy Factorial ---")
    try:
        num = int(input("Enter a positive integer: "))
        if num < 0:
            print("Factorial does not exist for negative numbers.")
        elif num == 0:
            print("0! = 1")
        else:
            # 1. Create array: [1, 2, 3, ... num]
            numbers = np.arange(1, num + 1)
            
            # 2. Multiply all elements in the array
            # We use dtype='object' to handle massive numbers (like 100!) 
            # that would otherwise crash a standard C-integer.
            result = np.prod(numbers, dtype='object')
            
            print(f"{num}! = {result}")
            
    except ValueError:
        print("Invalid input. Please enter an integer.")

# Driver code to test just this function
if __name__ == "__main__":
    get_factorial()
```
